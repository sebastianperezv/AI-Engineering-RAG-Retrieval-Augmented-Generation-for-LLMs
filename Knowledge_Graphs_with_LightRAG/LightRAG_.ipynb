Knowledge Graphs vs Embeddings

# Embeddings and Knowledge Graphs

Embeddings are mathematical representation of the data and we find what is closest to our query. When it comes to Knowledge graphs we find entities tat conects with each other.

## Quick Embedding recap

Mathematical representatnion of the data, they work in a high dimension vector space, capturing their meanings and similarities.

## Knowledge graphs

They organize data into nodes and edges, forming a clear network of relationships and attributes.
Knowledge graphs organize data into nodes and edges, forming a clear network of relationships and attributes.

## Key Difference

1. Nature of Relationships

> Embeddings: Relationships are implicit, inferred from emmbedding proximity and patterns like word co-occurence.

> Knowledge Graph: The relationships are explicit.

2. Interpretability

> Embeddings: Opaque, as measuring distance or angle between vectors shows similarity but is hard to interpret.

> Knowledge Graph: They are clear and easily interpretable.

3. Type of data Represented

> Embeddings: Great for capturing, similarity and context, excel at identifying items by usage patterns.

> Knowledge Graphs: Great for representing facts and relationships, they answer complex queries about entity connections.

Nodes:

- King Arthur
- Queen Guinevere
- Camelot
- Excalibur

Edges:

- King Arthur married to Quee Guinevere
- King Arthur rules Camelot
- Excalibur owned by King Arthur

What is LightRAG?

# LightRAG

This is a graph method that uses a graph best approach to store and retrieve chunks of information before passing to the LLM. It is like a tecnique that retrieves relevants facts when you try to answer a tricky question.

How LightRAG differs from Normal Knowledge Graphs.

1. Chunk-Centric vs Triple-Centric:

LightRAG adds text to nodes and edges, giving LLM context like how, why, and where events occurred.

2. Built for LLMs:

LightRAG focuses on retrival + generation with LLms, combining entity-edge data and text for natural language answers

3. Dual-level Retrival Triple-Centric:

LightRAG offers two retrival modes: one for precise facts and another for broader themes

4. Incremental updates:

LightRAG streamlines adding documents without a rebuild, keeping the graph updated effortlessly.

## Walking through LightRAG

1. Chunking text

LightRAG divides large documents into smaller chunks, ensuring efficient retrieval by avoiding full-document searches.

2. Entity & Relationship Extraction

LightRAG uses a language model to identify entities and their connections in each chunk, forming graph's nodes and edges.

3. Building the Knowledge Graph

LightRAG merges nodes, removes duplicates, and creates a graph with interconnections, keeping text snippets for LLM use.

4. Query Time - Dual-Level Retrival

* Low-level: Like Targeting 'Hermes' for related content. Focuses on the chunk matching the query.

* High level: Like pulling in Zeus and Hera when asking about 'Olympian gods'. Retrieves related nodes

5. Combine and Answer

LightRAG gives LLM curated text chunks and context for coherent, rich response.

## Summarize

LightRAG focuses on simplicity, speed and richer context, while traditional knowledge graphs excel at storing facts.

LightRAG combines a graph for connections with text chunks that provide the LLM exactly with what it needs.

Data Prep

# Data preparation

* PDF -> TXT (LightRag is for TXT)
* Create RAG with TXT

```python
# Open the PDF and extract the text
pdf_text = []
with open('Greek-gods-story.pdf', 'rb') as file:
  reader = PyPDF2.PdfReader(file)
  for page in reader.pages:
    pdf_text.append(page.extract_text())
```

```python
# what we currently have
pdf_text
```

```python
# Save (write) a file with the data
with open('Greek-gods-story.txt', 'w') as file:
  for line in pdf_text:
    file.write(f'{line}\n')
```

```python
# Build the RAG
with open('Greek-gods-story.txt', 'r') as file:
  rag.insert(file.read()) #We read the file and then we RAG it
```

Naive RAG

# RAG

```python
# Perform Naive RAG
def perform_rag(query, mode):
  return Markdown(rag.query(query, param = QueryParam(mode = mode)))

query = 'Who is Hermes?'
perform_rag(query, 'naive')
```

Implementing LightRAG

```python
# Perform LightRag with a local search
# This means we look into closer entities
perform_rag(query, 'local')
```

```python
# Perform LightRAG with global Search
# More diverse results, global entity relationships
perform_rag(query, 'global')
```

```python
# Perform LightRAG with Hybrid approach
# This more specific, when you really want to zoom in on the direct relationship
# or more high level relation ship
perform_rag(query, 'hybrid')
```

Knowledge Graph Visualization

# Graph Visualization

```python
# Import libraries
import networkx as nx
from pyvis.network import Network
import matplotlib.pyplot as plt
import random
```

```python
# Load entities
G = nx.read_graphml(f'{WORKING_DIR}/graph_chunk_entity_relation.graphml')

# Create a network
net = Network(notebook = True, cdn_resources = 'remote')
net.from_nx(G)

# Add colors and titles to nodes

for node in net.nodes:
  node['color'] = '#{:06x}'.format(random.randint(0, 0xFFFFFF))
  if 'description' in node:
    node['title'] = node['description']

# Add titles to edges
for edge in net.edges:
  if 'description' in edge:
    edge['title'] = edge['description']

# Save and display the network
net.show('graph.html')
net.save_graph('graph.html')

HTML(filename = 'graph.html')
```

```python
# Target SOCIEDAD SERVICIOS INGENIERÍA Y MANTENCIÓN (SIGMA) S.A.
# Load entities
G = nx.read_graphml(f'{WORKING_DIR}/graph_chunk_entity_relation.graphml')

# Target SOCIEDAD SERVICIOS INGENIERÍA Y MANTENCIÓN (SIGMA) S.A.
target_node = '"ZEUS"'

# Get neighbors for that node
subgraph_nodes = list(G.neighbors(target_node)) + [target_node]
subgraph = G.subgraph(subgraph_nodes)

# Create a network
net = Network(notebook = True, cdn_resources = 'remote')
net.from_nx(subgraph)

# Save and display the network
graph_name = f'{target_node}_graph.html'
net.save_graph(graph_name)
HTML(filename = graph_name)
```
